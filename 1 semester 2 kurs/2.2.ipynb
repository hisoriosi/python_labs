```python
import matplotlib.pyplot as plt
import numpy as np

# Открытие файла на считывание
f = open('var9.csv','r')

# Считывание и отбрасывание первых 16-и строк файла
for k in range(16):
  f.readline()

# Считывание массива данных
ch = []
for k in f:
  # Разделение считанной строки на части по символу ','
  s = k.split(',')
  # Преобразование первой части каждой строки (до запятой)
  # в число и добавление этого числа в массив       
  ch.append(int(s[0]))

ch = np.array(ch) # Преобразование типа из list в array
f.close() # Закрываем файл

# Параметры АЦП, полученные из заголовка
# и из паспортных данных
N = 10000          # Общее число точек сигнала
n = 8             # Разрядность АЦП
# Расчет диапазона входного сигнала
# Через перемножение числа делений на цену деления
# Цена деления 0,5В/дел, всего делений 10 (+5 и -5)
Smax = 5*0.5
Smin = -5*0.5
fs = 100000       # Частота дискретизации

h = 1/fs          # Шаг дискретизации

# Генерация массива координат точек по оси t
t = np.arange(N)*h

# Декодирование сигнала (беззнаковый код)
s = ch/(2**n-1)*(Smax-Smin) + Smin

# Построение графика считанного сигнала
plt.figure(1)
plt.plot(t,s)
plt.grid()
plt.title('График считанного из CSV файла сигнала')
plt.show(block=False)
```
