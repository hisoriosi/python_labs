```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Открытие файла
f = open('var9.csv', 'r', encoding='utf-8')

# 2. Считывание и отбрасывание строк заголовка
# (уточните количество строк заголовка в вашем файле)
for k in range(16):
    f.readline()

# 3. Считывание массива данных
ch = []
for k in f:
    s_line = k.strip().split(',')
    if s_line and s_line[0]:
        ch.append(int(s_line[0]))
ch = np.array(ch)
f.close()

# 4. Параметры АЦП
n = 8              # Разрядность АЦП
Smax = 5 * 0.5     # Максимальное напряжение
Smin = -5 * 0.5    # Минимальное напряжение
fs = 100000        # Частота дискретизации

# 5.  ВАЖНО: берём N из реального количества считанных точек
N = len(ch)
print(f"Количество считанных точек: {N}")

# 6. Расчет временной оси
h = 1 / fs
t = np.arange(N) * h

# 7. Декодирование сигнала (беззнаковый код)
s = ch / (2**n - 1) * (Smax - Smin) + Smin

# 8. Построение графика
plt.figure(1, figsize=(10, 5))
plt.plot(t, s)
plt.grid(True)
plt.title('График считанного из CSV файла сигнала (Вариант 9)')
plt.xlabel('Время, с')
plt.ylabel('Напряжение, В')
plt.show()
```
