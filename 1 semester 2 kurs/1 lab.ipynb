```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import sawtooth

# Данные варианта
f = 75
n = 8
Smin, Smax = -2.0, 2.0

# Расчет дискретизации
fs = 160 * f
h = 1 / fs
N = 320
t = np.arange(N) * h

# Генерация трапеции (tau=T/8)
raw = sawtooth(2 * np.pi * f * t, width=0.5)
s = np.clip(raw * 3.0, -1.5, 1.5)

# График
plt.plot(t, s)
plt.grid()
plt.title('Непрерывный сигнал')
plt.show()
```

```python
plt.stem(t, s, linefmt='r--', markerfmt='ro', basefmt='k-')
plt.grid()
plt.title('Дискретный сигнал')
plt.show()
```

```python
skod = np.floor((s - Smin) / (Smax - Smin) * (2**n - 1))

plt.step(t, skod, where='post', color='g')
plt.grid()
plt.title('Квантованный сигнал (коды)')
plt.show()
```

```python
s_kvant = skod / (2**n - 1) * (Smax - Smin) + Smin

plt.plot(t, s, label='Непрерывный')
plt.step(t, s_kvant, where='post', label='Квантованный')
plt.grid()
plt.legend()
plt.title('Совмещенный график')
plt.show()
```
